## 1. Motivation: photos are 2D, the world is 3D

An image is a **projection** of the 3D world from one viewpoint, with depth flattened away. After looking at a few photos, humans can "complete" the scene's 3D structure in their heads and imagine what it would look like from a new angle. The machine version of this ability is called **novel view synthesis**:

> **Input**: N training images + the camera pose of each
> **Output**: a rendered image at any new pose

The idea of **NeRF** (Mildenhall et al., 2020) is strikingly elegant: reconstruct neither a mesh nor a point cloud, but represent the scene itself directly with an MLP:

$$F_\theta: (x, y, z, \theta, \phi) \;\longrightarrow\; (\sigma, \, \mathrm{rgb})$$

i.e. at any point in space, the **density** (how much "stuff" is here) and the **colour** (what colour it looks from a given direction). Rendering an image = cast one ray per pixel, sample points along the ray, query the MLP, and integrate with **volume rendering**. Because the whole chain is differentiable, backpropagating the pixel error of "render vs real photo" gradually "condenses" the MLP into the scene's 3D structure.

**3D Gaussian Splatting (3DGS)** (Kerbl et al., 2023) swaps in an explicit representation for the same problem: the scene = millions of 3D Gaussians, each with a position, covariance, colour and opacity; rendering projects ("splats") them onto the image plane and α-blends them. It gives up the MLP's compactness in exchange for **real-time rendering**.

### ⚠️ An important distinction: this is a spatial world representation, not a complete world model

| | NeRF / 3DGS | the world models of Labs 2–4 |
|---|---|---|
| What it represents | the **geometry and appearance** of a static scene | how the state evolves under **actions** |
| Input | camera pose (passive observation) | state + **action** |
| Can answer | "what does it look like from here?" | "what happens if I do this?" |
| Time dimension | ❌ none | ✅ has dynamics |

NeRF/3DGS represent a **static, passively observed world**: they don't know what an action is and cannot predict the future. A complete world model needs both: **a spatial representation** (what the world looks like) × **dynamics** (how the world changes). Recent work combining 3DGS with dynamics (e.g. 4D Gaussians, Gaussian-based world simulators) is exactly where these two routes merge. By the end of this lab you will know where that boundary lies.

## 3. Environment & Dataset

### 3.1 Tiny synthetic scene: five coloured balls + an analytic renderer

To avoid depending on any external data, we put a toy scene in code: **5 spheres with different colours, positions and radii**, lit by one fixed directional light.

The renderer is **analytic**: cast one ray per pixel, solve the ray-sphere intersections, take the nearest hit, and shade it with Lambertian diffuse + a little view-dependent specular highlight; pixels that hit nothing get the background colour.

Note that **the specular term is added on purpose**: it makes the same 3D point look different from different directions, which is exactly why NeRF takes the viewing direction as input (Section 4.2).

```python
# ---- scene: (center, radius, albedo) ----
spheres = [
    (np.array([ 0.0,  0.0,  0.0]), 0.35, np.array([0.90, 0.15, 0.15])),  # big red ball in the middle
    (np.array([ 0.5,  0.1,  0.2]), 0.22, np.array([0.15, 0.35, 0.95])),
    (np.array([-0.4,  0.3, -0.1]), 0.20, np.array([0.15, 0.85, 0.25])),
    (np.array([ 0.1, -0.4,  0.3]), 0.18, np.array([0.95, 0.75, 0.10])),
    (np.array([-0.2, -0.2, -0.4]), 0.15, np.array([0.75, 0.20, 0.85])),
]
LIGHT = np.array([0.5, 0.8, 0.3]); LIGHT /= np.linalg.norm(LIGHT)
BG_COLOR = 0.92  # light gray background

def render_analytic(rays_o, rays_d):
    """analytic rendering: rays_o/rays_d (..., 3) -> rgb (..., 3). ray-sphere intersection + Lambert + specular."""
    img = np.full(rays_o.shape[:-1] + (3,), BG_COLOR, dtype=np.float32)
    t_best = np.full(rays_o.shape[:-1], np.inf, dtype=np.float32)
    for c, r, alb in spheres:
        oc = rays_o - c
        b = np.sum(oc * rays_d, axis=-1)          # quadratic coefficients (a=1, directions are normalized)
        disc = b * b - (np.sum(oc * oc, axis=-1) - r * r)
        t = -b - np.sqrt(np.maximum(disc, 0.0))
        hit = (disc > 0) & (t > 1e-4) & (t < t_best)
        if hit.any():
            p = rays_o[hit] + t[hit, None] * rays_d[hit]      # hit point
            n = p - c; n /= np.linalg.norm(n, axis=-1, keepdims=True)  # normal
            diff = np.clip(n @ LIGHT, 0, 1)[:, None]                   # Lambert
            refl = 2 * (n @ LIGHT)[:, None] * n - LIGHT
            spec = np.clip(np.sum(refl * (-rays_d[hit]), axis=-1), 0, 1)[:, None] ** 16 * 0.5
            img[hit] = (alb[None, :] * (0.25 + 0.75 * diff) + spec).astype(np.float32)
        t_best = np.where(hit, t, t_best)
    return np.clip(img, 0.0, 1.0)

print("scene ready:", len(spheres), "spheres")
```

### 3.2 Camera: pinhole model + known poses

**Pinhole camera**: focal length `f` (set by the FOV); the ray direction of pixel `(i, j)` in camera coordinates is `[(i - W/2)/f, -(j - H/2)/f, -1]`, which the **cam2world rotation matrix** then maps into the world frame.

Camera poses are built with **look-at**: the camera moves on a sphere of radius 3 around the scene, always looking at the origin.

- **Training views**: 30 of them, azimuths evenly spread around a full circle, elevation varying gradually between 11° and 32° (covering a band of viewpoints rather than a single ring; a single ring generalizes very poorly to new elevations);
- **Test views (novel views)**: 6 of them, azimuths in between the training azimuths, elevation 26°. The model **never sees them during training**.

The image resolution is deliberately 48×48: enough to see the five balls, and small enough for a CPU to train a NeRF in a few minutes.

```python
H = W = 48
FOCAL = 0.5 * W / np.tan(0.5 * np.deg2rad(40))
CAM_RADIUS = 3.0

def look_at(eye, target=np.zeros(3), up=np.array([0.0, 1.0, 0.0])):
    """returns the cam2world rotation matrix (3,3): the columns are the camera's right / up / -forward."""
    f = target - eye; f /= np.linalg.norm(f)
    r = np.cross(f, up); r /= np.linalg.norm(r)
    u = np.cross(r, f)
    return np.stack([r, u, -f], axis=1)

def make_pose(theta, elev, radius=CAM_RADIUS):
    """azimuth theta, elevation elev (radians) -> (eye, cam2world)"""
    eye = radius * np.array([np.cos(elev) * np.cos(theta),
                             np.sin(elev),
                             np.cos(elev) * np.sin(theta)])
    return eye, look_at(eye)

def get_rays(eye, c2w, H=H, W=W, focal=FOCAL):
    """pinhole camera: one ray per pixel -> rays_o, rays_d (H, W, 3), unit directions."""
    i, j = np.meshgrid(np.arange(W), np.arange(H), indexing="xy")
    dirs = np.stack([(i - W / 2 + .5) / focal,
                     -(j - H / 2 + .5) / focal,
                     -np.ones_like(i)], axis=-1)
    rd = np.sum(dirs[..., None, :] * c2w, axis=-1)
    rd /= np.linalg.norm(rd, axis=-1, keepdims=True)
    ro = np.broadcast_to(eye, rd.shape).copy()
    return ro.astype(np.float32), rd.astype(np.float32)

# ---- generate train / test data (analytic rendering, no download) ----
N_TRAIN, N_TEST = 30, 6
train_thetas = np.linspace(0, 2 * np.pi, N_TRAIN, endpoint=False)
train_elevs  = np.linspace(np.deg2rad(11), np.deg2rad(32), N_TRAIN)
test_thetas  = train_thetas[:N_TEST] + np.pi / N_TRAIN   # azimuths in between the training views
TEST_ELEV    = np.deg2rad(26)

train_views, test_views = [], []
for th, el in zip(train_thetas, train_elevs):
    eye, c2w = make_pose(th, el)
    ro, rd = get_rays(eye, c2w)
    train_views.append(dict(eye=eye, c2w=c2w, ro=ro, rd=rd, rgb=render_analytic(ro, rd)))
for th in test_thetas:
    eye, c2w = make_pose(th, TEST_ELEV)
    ro, rd = get_rays(eye, c2w)
    test_views.append(dict(eye=eye, c2w=c2w, ro=ro, rd=rd, rgb=render_analytic(ro, rd)))

print(f"train views: {len(train_views)} | test (novel) views: {len(test_views)} | image: {H}x{W}")
```

A look at the training data: 30 "photos" of 48×48, which is **all** the information NeRF gets (plus each photo's camera pose). Notice that the highlights move between viewpoints (view-dependent) and the occlusions between balls change too (parallax).

## 4. Build the Model

### 4.1 Positional encoding: teaching an MLP high-frequency detail

A counter-intuitive fact: feed `(x, y, z)` directly to an MLP and it **cannot learn sharp boundaries**. ReLU MLPs have a spectral bias and naturally prefer low-frequency functions (Rahaman et al., 2019), so the render smears into a blur.

NeRF's fix is to map every coordinate to a set of sin/cos at different frequencies:

$$\gamma(p) = \big[\sin(2^0\pi p), \cos(2^0\pi p), \dots, \sin(2^{L-1}\pi p), \cos(2^{L-1}\pi p)\big]$$

The frequencies are controlled by `L`: the larger `L`, the sharper the details it can express, but also the easier it overfits to the number of views. We use `L_pos=6` (position) and `L_dir=4` (direction).

```python
def posenc(x, L):
    """gamma(x): (..., C) -> (..., C + 2*C*L), frequencies 2^0 .. 2^(L-1)."""
    out = [x]
    for l in range(L):
        f = (2.0 ** l) * np.pi
        out += [torch.sin(f * x), torch.cos(f * x)]
    return torch.cat(out, dim=-1)

L_POS, L_DIR = 6, 4
IN_POS = 3 + 2 * 3 * L_POS   # 39
IN_DIR = 3 + 2 * 3 * L_DIR   # 27

# for intuition: a 1-D input x becomes a set of basis functions at different frequencies after posenc
xs = torch.linspace(-1, 1, 400)[:, None]
fig, ax = plt.subplots(figsize=(8, 2.5))
for l in range(L_POS):
    ax.plot(xs[:, 0], torch.sin((2.0 ** l) * np.pi * xs)[:, 0], label=f"$2^{l}\pi x$")
ax.set(title="positional encoding basis functions (sin components)",
       xlabel="x", ylabel="$\sin(2^l \pi x)$")
ax.legend(ncol=6, fontsize=8)
plt.tight_layout()
plt.savefig("assets/posenc_basis.png", dpi=130)
plt.show()
```

### 4.2 TinyNeRF: density looks only at position, colour also looks at direction

The network has two parts (a key design of the original NeRF, not an arbitrary choice):

1. **trunk**: takes the positionally encoded position `γ(x)` and outputs the **density σ** and a feature vector;
2. **colour head**: takes the feature + the positionally encoded direction `γ(d)` and outputs **rgb**.

Why split it this way?
- **σ (geometry) must not depend on the viewing direction**: whether a point is "solid" doesn't depend on where you stand. Letting σ see only `x` hard-codes this physical constraint into the network;
- **Colour can (and should) depend on direction**: highlights and specular reflections are view-dependent (remember the specular term deliberately added in 3.1).

The size is deliberately small: a two-layer 96-wide trunk and a one-layer 48-wide colour head, about 15k parameters in total (the original NeRF is 8×256).

```python
class TinyNeRF(nn.Module):
    """F_theta: (position, direction) -> (sigma, rgb)"""
    def __init__(self, hidden=96):
        super().__init__()
        self.trunk = nn.Sequential(
            nn.Linear(IN_POS, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden), nn.ReLU(),
        )
        self.sigma_head = nn.Linear(hidden, 1)          # density: position only
        self.color_head = nn.Sequential(                # colour: position features + viewing direction
            nn.Linear(hidden + IN_DIR, hidden // 2), nn.ReLU(),
            nn.Linear(hidden // 2, 3),
        )

    def forward(self, x, d):
        """x, d: (N, 3) -> sigma (N,1), rgb (N,3)"""
        h = self.trunk(posenc(x, L_POS))
        sigma = torch.relu(self.sigma_head(h))                       # sigma >= 0
        rgb = torch.sigmoid(self.color_head(torch.cat([h, posenc(d, L_DIR)], dim=-1)))
        return sigma, rgb

model = TinyNeRF()
n_params = sum(p.numel() for p in model.parameters())
print(f"TinyNeRF parameters: {n_params:,}  (original NeRF: ~1.2M)")
```

### 4.3 Volume rendering: integrating the points on a ray into a pixel colour

For a pixel, the camera casts a ray `r(t) = o + t·d`, samples `N` points uniformly in `[near, far]`, queries the MLP for each point's `(σ_i, c_i)`, and then computes the **volume rendering integral** (in discrete form):

$$\hat C(r) = \sum_i T_i\, \alpha_i\, c_i, \qquad \alpha_i = 1 - e^{-\sigma_i \delta}, \qquad T_i = \prod_{j<i}(1 - \alpha_j)$$

Intuition:
- `α_i` is the probability that the ray is **absorbed** in segment `i` (the larger σ and the step δ, the more likely it hits something);
- `T_i` (**transmittance**) is the probability that the ray **survives to segment i**, unblocked all the way;
- The final colour = each segment's colour weighted by the probability of "stopping here"; whatever is absorbed by no segment `(1 - Σw)` is filled with the background colour.

The same weights `w_i = T_i α_i` also give a **depth map**: `depth = Σ w_i t_i` (the expected termination depth), used in Section 6.2.

The whole chain is just additions, multiplications, exp and an MLP: **fully differentiable**, which is why NeRF can be trained with SGD.

```python
NEAR, FAR, N_SAMPLES = 2.0, 4.0, 32   # camera radius 3, scene near the origin -> the [2,4] segment of each ray covers every ball
T_VALS = torch.linspace(NEAR, FAR, N_SAMPLES)
DELTA = (FAR - NEAR) / N_SAMPLES

def render_rays(model, ro, rd):
    """NeRF volume rendering. ro, rd: (B, 3) -> rgb (B,3), depth (B,), acc (B,)"""
    B = ro.shape[0]
    pts = ro[:, None, :] + rd[:, None, :] * T_VALS[None, :, None]      # (B, N, 3) samples along the ray
    dirs = rd[:, None, :].expand_as(pts)
    sigma, rgb = model(pts.reshape(-1, 3), dirs.reshape(-1, 3))        # query the MLP
    sigma = sigma.reshape(B, N_SAMPLES)
    rgb = rgb.reshape(B, N_SAMPLES, 3)

    alpha = 1.0 - torch.exp(-sigma * DELTA)                            # absorption probability per segment
    trans = torch.cumprod(                                             # T_i: probability of surviving to segment i
        torch.cat([torch.ones_like(alpha[:, :1]), 1.0 - alpha + 1e-10], dim=1), dim=1)[:, :-1]
    w = alpha * trans                                                  # (B, N) weight per segment
    acc = w.sum(dim=1)                                                 # total probability the ray is absorbed
    rgb_out = (w[..., None] * rgb).sum(dim=1) + (1.0 - acc[:, None]) * BG_COLOR
    depth = (w * T_VALS[None, :]).sum(dim=1) / (acc + 1e-8)            # expected termination depth
    return rgb_out, depth, acc

print("render_rays ready — one ray:", N_SAMPLES, "samples, integration range [%.1f, %.1f]" % (NEAR, FAR))
```

## 5. Train

The training loop is so plain it hardly looks like a 3D reconstruction algorithm:

1. Flatten all pixel rays of the 30 training images into one pool (30 × 48 × 48 ≈ 69k rays);
2. At every step, draw 1024 random rays, render their colours with NeRF, and compute the **MSE** against the analytic renderer's ground truth;
3. Backpropagate with Adam.

No mesh, no point cloud, no explicit geometry: **the gradient is the only channel of information**. Every pixel votes on "what density and colour my ray should have", and over tens of thousands of votes the MLP gradually condenses the scene's 3D structure.

We also log the training PSNR (Peak Signal-to-Noise Ratio, `−10·log₁₀(MSE)`, higher is better; >20 dB is roughly recognizable, >25 dB is quite clear). 3000 steps take about 1–2 minutes on CPU.

```python
def stack_views(views):
    """flatten several views into a ray pool: ro, rd, rgb, each a (n_views*H*W, 3) torch tensor."""
    ro = torch.from_numpy(np.concatenate([v["ro"].reshape(-1, 3) for v in views]))
    rd = torch.from_numpy(np.concatenate([v["rd"].reshape(-1, 3) for v in views]))
    rgb = torch.from_numpy(np.concatenate([v["rgb"].reshape(-1, 3) for v in views]))
    return ro, rd, rgb

def train_nerf(views, n_iters=3000, batch=1024, lr=5e-3, log_every=100, verbose=True):
    """train a TinyNeRF on the given set of views; returns (model, history)."""
    ro, rd, rgb = stack_views(views)
    n_rays = ro.shape[0]
    model = TinyNeRF()
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    sched = torch.optim.lr_scheduler.ExponentialLR(opt, gamma=0.1 ** (1.0 / n_iters))
    hist = []
    t0 = time.time()
    for it in range(1, n_iters + 1):
        idx = torch.randint(0, n_rays, (batch,))
        pred, _, _ = render_rays(model, ro[idx], rd[idx])
        loss = ((pred - rgb[idx]) ** 2).mean()
        opt.zero_grad(); loss.backward(); opt.step(); sched.step()
        if it % log_every == 0:
            psnr = -10.0 * np.log10(loss.item())
            hist.append((it, loss.item(), psnr))
            if verbose and it % (log_every * 5) == 0:
                print(f"iter {it:5d} | loss {loss.item():.5f} | PSNR {psnr:5.1f} dB | {time.time()-t0:.0f}s")
    return model, hist

model, hist = train_nerf(train_views, n_iters=3000)
print(f"done. final train PSNR: {hist[-1][2]:.1f} dB")
```

### 6.2 Rendering any viewpoint: train view vs novel view vs ground truth

With the trained `model`, rendering a full image means: generate rays for every pixel of that view → `render_rays` → reshape. Below we define `render_view` and compare three kinds of images:

- **Training-view reconstruction** (an angle the model "has seen"): should be nearly perfect;
- **Novel view** (an azimuth + elevation never seen): the real test of novel view synthesis;
- The analytic **ground truth** renders of both, for reference.

```python
def render_view(model, eye, c2w, chunk=4096):
    """render a full image + depth map with NeRF. Returns rgb (H,W,3), depth (H,W), acc (H,W)."""
    model.eval()
    ro, rd = get_rays(eye, c2w)
    ro = torch.from_numpy(ro.reshape(-1, 3)); rd = torch.from_numpy(rd.reshape(-1, 3))
    rgbs, depths, accs = [], [], []
    with torch.no_grad():
        for s in range(0, ro.shape[0], chunk):
            rgb, dep, acc = render_rays(model, ro[s:s+chunk], rd[s:s+chunk])
            rgbs.append(rgb); depths.append(dep); accs.append(acc)
    return (torch.cat(rgbs).numpy().reshape(H, W, 3),
            torch.cat(depths).numpy().reshape(H, W),
            torch.cat(accs).numpy().reshape(H, W))

def psnr(a, b):
    return -10.0 * np.log10(np.mean((np.clip(a, 0, 1) - b) ** 2))

# 2 training views + 3 novel views
show = [("train", train_views[0]), ("train", train_views[15]),
        ("novel", test_views[1]), ("novel", test_views[3]), ("novel", test_views[5])]
fig, axs = plt.subplots(len(show), 3, figsize=(8, 2.1 * len(show)))
for row, (kind, v) in enumerate(show):
    pred, dep, acc = render_view(model, v["eye"], v["c2w"])
    p = psnr(pred, v["rgb"])
    axs[row, 0].imshow(v["rgb"]); axs[row, 0].set_ylabel(kind, fontsize=10)
    axs[row, 1].imshow(np.clip(pred, 0, 1))
    axs[row, 2].imshow(np.clip(np.abs(np.clip(pred, 0, 1) - v["rgb"]) * 3, 0, 1))  # error magnified 3x
    for c, t in enumerate(["ground truth", f"NeRF render ({p:.1f} dB)", "|error| x3"]):
        if row == 0: axs[row, c].set_title(t, fontsize=10)
    for c in range(3): axs[row, c].set_xticks([]); axs[row, c].set_yticks([])
plt.suptitle("NeRF: train-view reconstruction (top) vs novel view synthesis (bottom)", y=1.0)
plt.tight_layout()
plt.savefig("assets/novel_views.png", dpi=120, bbox_inches="tight")
plt.show()
```

### 6.3 Depth maps: geometry hidden in σ

We **never gave the model any depth supervision**, only RGB. But the volume rendering weights `w_i` automatically give each ray's expected termination depth `Σ w_i t_i`: multi-view consistency forces σ to peak at the true surfaces, so geometry emerges "for free".

Below, depth maps are shown next to the RGB renders (in the background `acc≈0`, depth is meaningless and shown as the far end).

## Advanced Extension (optional): from NeRF to 3D Gaussian Splatting

NeRF uses an **implicit** representation (the scene lives in the MLP's weights), and rendering one pixel requires dozens of network queries: slow. **3DGS** switches to an **explicit** representation: the scene = a set of 3D Gaussian ellipsoids, each with a position μ, a covariance Σ (rotation + scale), a colour and an opacity α. To render, each 3D Gaussian is **projected (splatted) onto the image plane** as a 2D Gaussian, and after sorting by depth they are **α-blended**:

$$C = \sum_i c_i\, \alpha_i\, G_i^{2D}(p) \prod_{j<i}\big(1 - \alpha_j G_j^{2D}(p)\big)$$

Formally this is identical to NeRF's volume rendering (transmittance × α × colour); it just replaces "sampling along a ray" with "flattening Gaussians onto the screen". All parameters are directly differentiable, and rendering uses rasterization instead of point-by-point queries, which is why it runs in real time (100+ FPS).

Below, a **simplified 2D splatting** demo shows the core mechanism: 8 hand-placed 2D Gaussians α-blended into one picture in depth order. The panels show what each Gaussian "looks like" as it is added, and the composited result.

```python
# ---- simplified 2D Gaussian Splatting: 8 hand-placed Gaussians alpha-blended into a picture ----
gaussians2d = [  # (mu_x, mu_y, sx, sy, rot, color, alpha, z)  larger z is drawn first (farther)
    (0.50, 0.55, 0.28, 0.28, 0.0,  [0.96, 0.85, 0.70], 0.95, 8),  # warm sky light
    (0.30, 0.62, 0.10, 0.16, 0.4,  [0.95, 0.95, 0.98], 0.80, 7),  # cloud
    (0.68, 0.60, 0.12, 0.10, -0.3, [0.95, 0.95, 0.98], 0.75, 6),
    (0.50, 0.38, 0.30, 0.16, 0.0,  [0.25, 0.45, 0.75], 0.90, 5),  # distant mountains
    (0.42, 0.30, 0.20, 0.10, 0.2,  [0.30, 0.60, 0.35], 0.90, 4),  # bushes
    (0.66, 0.28, 0.14, 0.08, -0.2, [0.25, 0.55, 0.30], 0.90, 3),
    (0.50, 0.12, 0.32, 0.14, 0.0,  [0.85, 0.75, 0.55], 0.95, 2),  # ground
    (0.44, 0.16, 0.05, 0.05, 0.0,  [0.90, 0.20, 0.20], 0.95, 1),  # red flower (nearest)
]

def eval_gaussian2d(grid, mu, sx, sy, rot):
    """evaluate an anisotropic Gaussian at the grid points -> (P,) unnormalized density."""
    c, s = np.cos(rot), np.sin(rot)
    R = np.array([[c, -s], [s, c]])
    prec = R @ np.diag([1 / sx**2, 1 / sy**2]) @ R.T          # precision matrix Sigma^-1
    d = grid - mu
    maha = np.einsum("pi,ij,pj->p", d, prec, d)
    return np.exp(-0.5 * maha)

P = 240
yy, xx = np.mgrid[0:P, 0:P]
grid = np.stack([xx.ravel() / P, 1 - yy.ravel() / P], axis=1)  # (P*P, 2), y points up

canvas = np.ones((P * P, 3), dtype=np.float32)                 # white background
layers = []
for mu_x, mu_y, sx, sy, rot, color, alpha, z in sorted(gaussians2d, key=lambda g: -g[7]):
    g = eval_gaussian2d(grid, np.array([mu_x, mu_y]), sx, sy, rot)
    a = (alpha * g)[:, None]
    layers.append((np.array(color) * a + canvas * (1 - a)))    # 'over' compositing
    canvas = layers[-1]

fig, axs = plt.subplots(2, len(layers) // 2 + 1, figsize=(13, 5.5))
for k, ax in enumerate(axs.flat):
    ax.axis("off")
    if k < len(layers):
        ax.imshow(layers[k].reshape(P, P, 3).clip(0, 1))
        ax.set_title(f"+ gaussian {k}", fontsize=9)
    elif k == len(layers):
        ax.imshow(canvas.reshape(P, P, 3).clip(0, 1))
        ax.set_title("composited", fontsize=10, fontweight="bold")
plt.suptitle("2D Gaussian splatting: back-to-front alpha-blending (the core mechanism of 3DGS)")
plt.tight_layout()
plt.savefig("assets/splatting_2d.png", dpi=120, bbox_inches="tight")
plt.show()
```

**The 2D demo only shows the compositing mechanism**: the Gaussians are placed by hand, not projected from 3D, and nothing is learned. The next section adds all three: on the same five-ball scene, with exactly the same data and loss as NeRF, we train a real 3D Gaussian Splatting.

## Advanced Extension B: train a tiny 3D Gaussian Splatting yourself

The 2D demo above was placed by hand. Now we turn it into real 3DGS: **the same scene, the same 30 training images, the same photometric MSE**, only replacing NeRF's "MLP + sampling along rays" with "a set of learnable 3D Gaussians + splatting".

**Parameters of each Gaussian** (all learned directly by gradient descent):

| Parameter | Shape | Meaning |
|---|---|---|
| position μ | 3 | Gaussian centre |
| scale s (stored as log) | 3 | standard deviations along the three principal axes |
| rotation q (quaternion) | 4 | orientation of the axes; $\Sigma = R\,S\,S^\top R^\top$ keeps the covariance positive definite |
| colour c (stored as logit) | 3 | view-independent here (real 3DGS uses spherical harmonics for view-dependent colour) |
| opacity α (stored as logit) | 1 | |

**Rendering = projection + sorting + α-blending**:

1. Transform the centres into camera coordinates and sort by depth $d$ (nearest first);
2. Project each 3D covariance to a 2D covariance on screen (EWA splatting): $\Sigma' = J\,W\,\Sigma\,W^\top J^\top$, where $W$ is the world→camera rotation and $J$ is the Jacobian of the perspective projection at that point;
3. Composite every pixel in depth order: $C = \sum_i c_i\,\alpha_i' \prod_{j<i}(1-\alpha_j') + T_{\text{final}}\cdot \text{bg}$, with $\alpha_i' = \alpha_i \exp\!\big(-\tfrac12 \Delta^\top \Sigma_i'^{-1}\Delta\big)$.

This matches NeRF's volume rendering formula term by term: transmittance × α × colour. The difference is that there is **no network and no sampling along rays**: each pixel only needs one analytic evaluation per Gaussian.

**Initialization**: real 3DGS initializes the Gaussian centres from a sparse point cloud reconstructed by COLMAP (SfM). We haven't run SfM, but we do have a trained NeRF, and its depth maps (Section 6.3) are ready-made geometry: back-project the depth of a few training views into a point cloud, pick N random points as Gaussian centres, and initialize their colours from the corresponding pixels.

```python
# ---- tiny 3DGS: learnable 3D Gaussians + a differentiable splatting renderer ----
def quat_to_rotmat(q):
    """(N,4) quaternions (w,x,y,z) -> (N,3,3) rotation matrices; normalized first, so q can be optimized freely."""
    q = q / q.norm(dim=-1, keepdim=True)
    w, x, y, z = q.unbind(-1)
    return torch.stack([1 - 2*(y*y + z*z), 2*(x*y - w*z),     2*(x*z + w*y),
                        2*(x*y + w*z),     1 - 2*(x*x + z*z), 2*(y*z - w*x),
                        2*(x*z - w*y),     2*(y*z + w*x),     1 - 2*(x*x + y*y)], -1).reshape(-1, 3, 3)

class TinyGaussians(nn.Module):
    def __init__(self, means, colors, init_scale=0.05):
        super().__init__()
        n = len(means)
        self.means = nn.Parameter(torch.as_tensor(means, dtype=torch.float32).clone())
        self.log_scales = nn.Parameter(torch.full((n, 3), float(np.log(init_scale))))
        self.quats = nn.Parameter(torch.tensor([[1.0, 0.0, 0.0, 0.0]]).repeat(n, 1))
        c = torch.as_tensor(colors, dtype=torch.float32).clamp(0.02, 0.98)
        self.color_logits = nn.Parameter(torch.log(c / (1 - c)))       # inverse of sigmoid
        self.opacity_logits = nn.Parameter(torch.full((n,), 1.0))      # sigmoid(1) ≈ 0.73

PIX_XY = torch.from_numpy(np.stack(np.meshgrid(np.arange(W), np.arange(H), indexing="xy"), -1)
                          .reshape(-1, 2).astype(np.float32))           # (H*W, 2) pixel coordinates (i, j)

def render_gaussians(gs, eye, c2w, bg=BG_COLOR):
    """Splat every Gaussian onto an (H, W) image, with the same pinhole convention as get_rays."""
    R = torch.from_numpy(c2w.astype(np.float32))                       # cam2world, columns = right / up / back
    p = (gs.means - torch.from_numpy(eye.astype(np.float32))) @ R      # world -> camera coordinates
    d = -p[:, 2]                                                       # depth (the camera looks down -z)
    keep = d > 0.1                                                     # keep only Gaussians in front of the camera
    p, d = p[keep], d[keep]
    S = torch.exp(gs.log_scales[keep])
    M = quat_to_rotmat(gs.quats[keep]) * S[:, None, :]                 # R @ diag(S)
    cov_cam = R.T @ (M @ M.transpose(1, 2)) @ R                        # 3D covariance in the camera frame
    J = torch.zeros(len(d), 2, 3)                                      # Jacobian of the perspective projection
    J[:, 0, 0] = FOCAL / d; J[:, 0, 2] = FOCAL * p[:, 0] / d**2
    J[:, 1, 1] = -FOCAL / d; J[:, 1, 2] = -FOCAL * p[:, 1] / d**2
    cov2d = J @ cov_cam @ J.transpose(1, 2) + 0.3 * torch.eye(2)       # +0.3: anti-aliasing low-pass filter, prevents degenerate splats
    mu2d = torch.stack([W / 2 - 0.5 + FOCAL * p[:, 0] / d,
                        H / 2 - 0.5 - FOCAL * p[:, 1] / d], -1)        # centres projected to pixel coordinates
    order = torch.argsort(d)                                           # near -> far
    mu2d, cov2d = mu2d[order], cov2d[order]
    color = torch.sigmoid(gs.color_logits[keep][order])
    opacity = torch.sigmoid(gs.opacity_logits[keep][order])
    a, b, c = cov2d[:, 0, 0], cov2d[:, 0, 1], cov2d[:, 1, 1]           # inverse of the 2x2 covariance, by hand
    det = a * c - b * b
    dx = PIX_XY[:, :1] - mu2d[None, :, 0]                              # (H*W, N)
    dy = PIX_XY[:, 1:] - mu2d[None, :, 1]
    maha = (c * dx * dx - 2 * b * dx * dy + a * dy * dy) / det         # squared Mahalanobis distance
    alpha = (opacity[None] * torch.exp(-0.5 * maha)).clamp(max=0.99)
    T = torch.cumprod(torch.cat([torch.ones(len(PIX_XY), 1), 1 - alpha[:, :-1]], 1), 1)  # transmittance
    w = alpha * T                                                      # same form as NeRF's w_i = T_i * alpha_i
    rgb = w @ color + (1 - w.sum(1, keepdim=True)) * bg                # whatever is not covered = background
    return rgb.reshape(H, W, 3)
print("render_gaussians ready")
```

```python
# ---- initialization: NeRF depth maps instead of an SfM point cloud ----
N_GAUSS = 300
pts, cols = [], []
for v in train_views[::3]:                                   # 10 training views are enough
    _, dep, acc = render_view(model, v["eye"], v["c2w"])
    m = acc > 0.9                                            # only pixels where NeRF is sure the ray "hit something"
    pts.append(v["ro"][m] + dep[m, None] * v["rd"][m])       # back-project: o + depth * d
    cols.append(v["rgb"][m])
pts, cols = np.concatenate(pts), np.concatenate(cols)
pick = rng.choice(len(pts), N_GAUSS, replace=False)
gs = TinyGaussians(pts[pick], cols[pick])
print(f"point cloud from NeRF depth: {len(pts)} points -> {N_GAUSS} gaussians "
      f"({sum(p.numel() for p in gs.parameters()):,} parameters)")

fig = plt.figure(figsize=(9, 3.4))
ax = fig.add_subplot(1, 2, 1, projection="3d")
ax.scatter(*pts[pick].T, c=cols[pick], s=8)
ax.set_title("initial gaussian centres (from NeRF depth)", fontsize=9)
with torch.no_grad():
    init_img = render_gaussians(gs, test_views[1]["eye"], test_views[1]["c2w"]).numpy()
ax2 = fig.add_subplot(1, 2, 2)
ax2.imshow(np.clip(init_img, 0, 1)); ax2.axis("off")
ax2.set_title("render before training (novel view)", fontsize=9)
plt.tight_layout(); plt.show()
```

```python
# ---- training: one random training view per step, photometric MSE over the whole image ----
GS_ITERS = 800
opt_gs = torch.optim.Adam([
    {"params": [gs.means],          "lr": 5e-3},
    {"params": [gs.log_scales],     "lr": 1e-2},
    {"params": [gs.quats],          "lr": 1e-2},
    {"params": [gs.color_logits],   "lr": 5e-2},
    {"params": [gs.opacity_logits], "lr": 5e-2},
])
gs_targets = [torch.from_numpy(v["rgb"]) for v in train_views]
gs_hist, t_gs = [], time.time()
for it in range(1, GS_ITERS + 1):
    k = int(rng.integers(len(train_views)))
    v = train_views[k]
    loss = ((render_gaussians(gs, v["eye"], v["c2w"]) - gs_targets[k]) ** 2).mean()
    opt_gs.zero_grad(); loss.backward(); opt_gs.step()
    gs_hist.append(loss.item())
    if it % 100 == 0:
        print(f"iter {it:4d} | loss {np.mean(gs_hist[-100:]):.5f} | {time.time() - t_gs:.0f}s")
print(f"3DGS training done in {time.time() - t_gs:.0f}s")
```

```python
# ---- evaluation: image quality (PSNR) and rendering speed, NeRF vs 3DGS ----
def render_gs_np(eye, c2w):
    with torch.no_grad():
        return render_gaussians(gs, eye, c2w).numpy()

gs_train_psnrs = np.array([psnr(render_gs_np(v["eye"], v["c2w"]), v["rgb"]) for v in train_views])
gs_novel_psnrs = np.array([psnr(render_gs_np(v["eye"], v["c2w"]), v["rgb"]) for v in test_views])

def time_render(fn, n=10):
    fn(); t = time.perf_counter()
    for _ in range(n): fn()
    return (time.perf_counter() - t) / n * 1000

v = test_views[0]
ms_nerf = time_render(lambda: render_view(model, v["eye"], v["c2w"]))
ms_gs = time_render(lambda: render_gs_np(v["eye"], v["c2w"]))
n_nerf = sum(p.numel() for p in model.parameters()); n_gs = sum(p.numel() for p in gs.parameters())
print(f"{'':6s} {'params':>8s} {'train PSNR':>11s} {'novel PSNR':>11s} {'ms / frame (CPU)':>17s}")
print(f"{'NeRF':6s} {n_nerf:8,d} {train_psnrs.mean():10.1f}  {novel_psnrs.mean():10.1f}  {ms_nerf:16.1f}")
print(f"{'3DGS':6s} {n_gs:8,d} {gs_train_psnrs.mean():10.1f}  {gs_novel_psnrs.mean():10.1f}  {ms_gs:16.1f}")

fig, axs = plt.subplots(3, 3, figsize=(7.5, 7.5))
for row, vv in enumerate([test_views[1], test_views[3], test_views[5]]):
    nerf_img = render_view(model, vv["eye"], vv["c2w"])[0]
    gs_img = render_gs_np(vv["eye"], vv["c2w"])
    for col, (img, name) in enumerate([(vv["rgb"], "ground truth"),
                                       (nerf_img, f"NeRF ({psnr(nerf_img, vv['rgb']):.1f} dB)"),
                                       (gs_img, f"3DGS ({psnr(gs_img, vv['rgb']):.1f} dB)")]):
        axs[row, col].imshow(np.clip(img, 0, 1)); axs[row, col].set_xticks([]); axs[row, col].set_yticks([])
        axs[row, col].set_title(name, fontsize=9)
plt.suptitle("novel views: implicit NeRF vs explicit 3D Gaussians", y=1.0)
plt.tight_layout()
plt.savefig("assets/gaussian_splatting_3d.png", dpi=120, bbox_inches="tight")
plt.show()
```

**Reading the results**: the two representations use **the same data and the same loss**, but they are completely different parameterizations. NeRF hides the scene in about 19k MLP weights; 3DGS spreads it out into a few hundred ellipsoids you can see and inspect one by one. 3DGS uses about a fifth of NeRF's parameters and gets the geometry and colours right, but its edges are softer and its image quality is a few dB lower: our Gaussians' colour ignores the viewing direction (NeRF's colour is view-dependent), their number is fixed with no densification, and training runs for only a few hundred steps. What it buys is rendering speed: 3DGS only needs one analytic evaluation per Gaussian, whereas NeRF queries the MLP 32 times per pixel, so even our naive CPU implementation renders a frame several times faster.

**What separates this tiny version from real 3DGS?**

① **Adaptive densification / pruning**: during training, clone or split Gaussians with large gradients and delete nearly transparent ones; we keep a fixed 300;
② **Spherical-harmonics colour**: colour that changes with the viewing direction, to express highlights;
③ **Tile-based rasterization**: the screen is cut into 16×16 tiles and each tile only processes the Gaussians that cover it, in parallel on CUDA. Our implementation evaluates every Gaussian at every pixel ($O(HW \cdot N)$), which is exactly why it would not scale to millions of Gaussians while the official implementation runs at 100+ FPS;
④ **SfM initialization**: COLMAP estimates the camera poses and a sparse point cloud from the photos themselves; we simply borrowed the known poses and NeRF's depth.

To run real 3DGS, we recommend:

- **Official implementation**: [graphdeco-inria/gaussian-splatting](https://github.com/graphdeco-inria/gaussian-splatting) (CUDA, needs a GPU);
- **Teaching-oriented integration**: `splatfacto` in [nerfstudio](https://docs.nerf.studio/), sharing a pipeline with NeRF; one command, `ns-train splatfacto --data ...`;
- **Lightweight reimplementation**: [gsplat](https://github.com/nerfstudio-project/gsplat) (a standalone CUDA library from the nerfstudio team).

All three are worth reading side by side with this lab's tiny NeRF and tiny 3DGS: you have already written the core formulas yourself.

## 10. Takeaways

- **NeRF = packing a whole scene into one MLP**: `F_θ(x, d) → (σ, rgb)`; rendering = ray sampling + the volume rendering integral; training = backpropagating a per-pixel photometric loss. With no explicit geometry at all, the 3D structure (depth maps!) emerges automatically from multi-view RGB constraints;
- **Two key designs**: positional encoding overcomes the MLP's low-frequency bias; σ looks only at position while colour also looks at direction, hard-coding the physical constraint "geometry doesn't depend on direction" into the network;
- **The quality of novel view synthesis is set by viewpoint coverage**: with too few training views, the model keeps a high PSNR on the training views while hiding foggy floaters in mid-air. The train/novel gap is the most honest diagnostic;
- **3DGS is another parameterization of the same integral**: explicit 3D Gaussians + splatting + α-blending, giving up implicit compactness in exchange for real-time rendering. The volume rendering formulas of NeRF and 3DGS are essentially the same. In Advanced Extension B you trained a tiny 3DGS with a few hundred Gaussians on the same data and loss: it has fewer parameters and renders each frame several times faster (only an analytic evaluation of the Gaussians, no network queries), at the cost of a few dB of image quality without densification or view-dependent colour;
- **Draw the boundary**: NeRF/3DGS are a **spatial world representation**, a representation of a static, passively observed world; they contain no actions and no dynamics, and cannot predict "what happens if I...". Combining them with the dynamics models of Labs 2–4 (4D representations, action-conditioned rendering, neural simulators you can plan in) is the path toward a complete spatial world model.
